```python
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings

emb = OpenAIEmbeddings(model="text-embedding-3-large")

db = Chroma(
    persist_directory="./rag_market9",
    embedding_function=emb
)

# 전체 컬렉션 정보 추출
collection = db._collection.get()

print(collection.keys())  # 어떤 필드 있는지 확인
print("📌 전체 문서 수:", len(collection["documents"]))

# 앞 3개 문서 내용 보기
for i in range(3):
    print("="*40)
    print("문서", i+1)
    print(collection["documents"][i])
    print("metadata:", collection["metadatas"][i])
```

```python
import pkg_resources
import re

# ✅ 대시보드에서 실제로 사용하는 주요 패키지
key_packages = [
    "streamlit", "pandas", "matplotlib",
     "joblib", "shap", "plotly", "langchain", "langgraph",
     "chromadb", "openai", "tiktoken", "streamlit_chat", "rag", "langchain-core","langchain-community","langchain_openai"]


# 현재 설치된 모든 패키지 목록 가져오기
installed = {pkg.key: pkg.version for pkg in pkg_resources.working_set}

# key_packages 중 설치되어 있는 것만 추출
filtered = {pkg: installed[pkg] for pkg in key_packages if pkg in installed}

# requirements.txt 형식으로 정리
lines = [f"{pkg}=={ver}" for pkg, ver in filtered.items()]
requirements_text = "\n".join(lines)

# ✅ 파일로 저장
with open("requirements.txt", "w", encoding="utf-8") as f:
    f.write(requirements_text)

print("✅ requirements.txt 생성 완료!")
print("------ 내용 미리보기 ------")
print(requirements_text)
```

```python
from langchain.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings

emb = OpenAIEmbeddings(model="text-embedding-3-large")
db_region = Chroma(
    persist_directory="./rag_region6",
    embedding_function=emb
)

collection = db_region._collection.get()

all_regions = [
    m.get("region")
    for m in collection["metadatas"]
]

print("=== DB에 저장된 region 값들 ===")
for r in set(all_regions):
    print(repr(r))
```

```python
docs_ydp = db_region.similarity_search(
    "전략",         # 아무 텍스트나 상관 없음
    k=5,
    filter={"region": "영등포구"}
)

print("영등포구 문서 개수:", len(docs_ydp))
for d in docs_ydp:
    print(d.metadata, d.page_content[:80], "...")
```
